Análisis de humedad de 5 lotes de cacao de PARAMANGLA. Requisito: humedad máxima 2 %.

```python
humedad = [1.6, 2.8, 1.9, 2.1, 1.7]

promedio = sum(humedad) / len(humedad)
print("Promedio:", round(promedio, 2))
print("Máximo:", max(humedad))
print("Mínimo:", min(humedad))
```

```python
fuera = 0
for h in humedad:
    if h > 2:
        fuera = fuera + 1

print("Lotes fuera de requisito:", fuera, "de", len(humedad))
print("Porcentaje fuera:", round(fuera / len(humedad) * 100, 1), "%")
```
